```python
import pandas as pd

df = pd.read_csv("../data/raw/2024.csv")

print(df.shape)
```

```python
print(list(df.columns))
df.head(3)
```

```python
#1
print(df["tourney_level"].value_counts(dropna=False))
print(df["round"].value_counts(dropna=False))
```

```python
#2
for col in ["surface", "indoor", "best_of"]:
    print(df[col].value_counts(dropna=False))
    print()
```

```python
#3
cols = ["minutes", "w_ace", "w_svpt", "l_ace", "l_svpt", "winner_rank", "loser_rank"]
print(df[cols].isna().mean().round(3))
```

```python
#4
print(df["tourney_date"].dtype, df["tourney_date"].min(), df["tourney_date"].max())
print(df.duplicated(subset=["tourney_id", "match_num"]).sum())
```

```python
#5
print(df["score"].str.contains("W/O", na=False).sum())
print(df["score"].str.contains("RET", na=False).sum())
```

```python
#6 
for lvl in ["A", "O", "F"]:
    print(lvl)
    print(df[df["tourney_level"] == lvl]["tourney_name"].value_counts())
    print()
```

```python

m = df[df["tourney_level"] == "M"]
print(m.groupby("tourney_name").agg(matches=("match_num", "count"), draw=("draw_size", "max")))
```

```python
print(pd.crosstab(df["tourney_level"], df["indoor"].fillna("missing")))
```

```python
print(df.duplicated(subset=["tourney_id", "match_num"]).sum())
print(df[(df["best_of"] == 5) & (df["tourney_level"] != "G")]["tourney_level"].value_counts())
```

```python
g = df[df["tourney_level"] == "G"]
print(g.groupby("tourney_name").agg(matches=("match_num", "count"), draw=("draw_size", "max")))
```
