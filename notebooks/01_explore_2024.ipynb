```python
import pandas as pd

df = pd.read_csv("../data/raw/2024.csv")

print(df.shape)
```

```python

m = df[df["tourney_level"] == "M"]
print(m.groupby("tourney_name").agg(matches=("match_num", "count"), draw=("draw_size", "max")))
```

```python
print(pd.crosstab(df["tourney_level"], df["indoor"].fillna("missing")))
```

```python
print(df.duplicated(subset=["tourney_id", "match_num"]).sum())
print(df[(df["best_of"] == 5) & (df["tourney_level"] != "G")]["tourney_level"].value_counts())
```

```python
g = df[df["tourney_level"] == "G"]
print(g.groupby("tourney_name").agg(matches=("match_num", "count"), draw=("draw_size", "max")))
```

```python
excluded_names = ["United Cup", "Laver Cup", "Next Gen Finals", "Paris Olympics"]

clean = df.copy()
print("start:", len(clean))

clean = clean[clean["tourney_level"] != "D"]
print("after removing Davis Cup:", len(clean))

clean = clean[~clean["tourney_name"].isin(excluded_names)]
print("after removing excluded events:", len(clean))

clean = clean[~clean["score"].str.contains("W/O", na=False)]
print("after removing walkovers:", len(clean))
```

```python
clean["date"] = pd.to_datetime(clean["tourney_date"].astype(str), format="%Y%m%d")
clean["retired"] = clean["score"].str.contains("RET", na=False)

print(clean["date"].min(), clean["date"].max())
print(clean["retired"].sum())
```

```python
print(clean["score"].str.findall(r"[A-Za-z/]+").explode().value_counts())
```

```python
core = clean[clean["tourney_level"].isin(["G", "M"])]
print(core.groupby("tourney_name")["match_num"].count())
print(len(core["tourney_name"].unique()))
```

```python
core = clean[clean["tourney_level"].isin(["G", "M", "F"])]
print(core.groupby("tourney_name")["match_num"].count())
print(len(core["tourney_name"].unique()))
```
